# 실습5

```python
import pandas as pd

aptSales = pd.read_csv("apt_sales.csv", encoding="utf-8")         #아파트 매매거래 정보
aptBasicInfo = pd.read_csv("apt_basicinfo.csv", encoding="utf-8") #아파트기본 정보
houseSales = pd.read_csv("house_sales.csv", encoding="utf-8")     #단독주택 매매거래 정보
houseInfo = pd.read_csv("house_basicinfo.csv", encoding="utf-8")  #단독주택 기본 정보
```

## 1번
- 1-1. 단독주택 매매거래 정보(houseSales) 데이터에 총 몇 개의 열이 저장되어 있는지
확인하는 코드를 작성하고, 그 중 매매가격(salesPrice) 변수가 몇 번째 열에 위치해 있는
지 확인하여 적으시오.

```python
print("houseSales 열 개수:", len(houseSales.columns))
print("salesPrice 변수 위치:", houseSales.columns.get_loc("salesPrice"))
```

- 1.2. id 열을 기준으로 매매거래 정보(houseSales)와 기본 정보(houseInfo) 데이터를 결
합하여 새로운 데이터프레임으로 저장하는 코드를 작성하시오.

```python
houseMerged = pd.merge(houseSales, houseInfo, on="id")
print(houseMerged.head())
```

- 1-3. 결합 후 생성된 데이터프레임에서 중복된 열 이름(예: si_x, si_y)이 있는 경우, 이 중
에서 임의로 하나를 제거하고 새로운 데이터프레임으로 저장하는 코드를 작성하시오.

```python
houseCleaned = houseMerged.drop(columns=["si_y", "gu_y", "dong_y", "bunji_y"])
print(houseCleaned.head())
```

- 1-4. 중복된 열을 제거한 후, 남아 있는 열 이름을 원래 열 이름으로 변경(예: si_x → si)
하고, 이를 새로운 데이터프레임으로 저장하는 코드를 작성하시오.

```python
houseRenamed = houseCleaned.rename(
    columns={"si_x": "si", "gu_x": "gu", "dong_x": "dong", "bunji_x": "bunji"}
)
print(houseRenamed.head())
```

- 1-5. 동대문구에 소재한 단독주택의 단위면적당(landSize) 매매가격에 대한 기초 통계량
을 확인하는 코드를 작성하시오.

```python
houseRenamed["pricePerSize"] = houseRenamed["salesPrice"] / houseRenamed["landSize"]
print(houseRenamed.loc[houseRenamed["gu"] == "동대문구", "pricePerSize"].describe())
```

- 1-6. 동대문구와 성북구 단독주택의 매매가격 평균을 각각 계산하여 출력하는 코드를
작성하시오.

```python
print("동대문구 매매가격 평균:", houseRenamed.loc[houseRenamed["gu"] == "동대문구", "salesPrice"].mean())
print("성북구 매매가격 평균:", houseRenamed.loc[houseRenamed["gu"] == "성북구", "salesPrice"].mean())
```

## 2번

- 2-1. 아파트 매매거래 정보(aptSales) 데이터에 총 몇 개의 열이 저장되어 있는지 확인
하는 코드를 작성하고, 그 중 매매가격(salesPrice) 변수가 몇 번째 열에 위치해 있는지
확인하여 적으시오.

```python
print("aptSales데이터 열 개수:", len(aptSales.columns))
print("salesPrice 변수 위치:", aptSales.columns.get_loc("salesPrice"))
```

- 2-2. id 열을 기준으로 매매거래 정보(aptSales)와 기본 정보(aptBasicInfo) 데이터를 결합
하여 새로운 데이터프레임으로 저장하는 코드를 작성하시오.


```python
aptMerged = pd.merge(aptSales, aptBasicInfo, on="id")
print(aptMerged.head())
```

- 2-3. 결합 후 생성된 데이터프레임에서 중복된 열 이름(예: si_x, si_y)이 있는 경우, 이 중
에서 임의로 하나를 제거하고 새로운 데이터프레임으로 저장하는 코드를 작성하시오.

```python
aptCleaned = aptMerged.drop(columns=["si_y", "gu_y", "dong_y", "bunji_y", "mainBun_y", "subBun_y"])
print(aptCleaned.head())
```

- 2-4. 중복된 열을 제거한 후, 남아 있는 열 이름을 원래 열 이름으로 변경(예: si_x → si)
하고, 이를 새로운 데이터프레임으로 저장하는 코드를 작성하시오.

```python
aptRenamed = aptCleaned.rename(
    columns={"si_x": "si", 
             "gu_x": "gu", 
             "dong_x": "dong", 
             "bunji_x": "bunji",
             "mainBun_x": "mainBun",
             "subBun_x": "subBun"
             }
)
print(aptRenamed.head())
```

- 2-5. 성북구에 소재한 아파트의 단위면적당(size) 매매가격에 대한 기초 통계량을 확인
하는 코드를 작성하시오.

```python
aptRenamed["pricePerSize"] = aptRenamed["salesPrice"] / aptRenamed["size"]
print(aptRenamed.loc[aptRenamed["gu"] == "성북구", "pricePerSize"].describe())
```

- 2-6. 동대문구와 성북구 아파트의 매매가격 평균을 각각 계산하여, 그 결과를 기존 데
이터프레임의 각 행에 대응되는 새로운 파생변수로 추가하는 코드를 작성하시오.

```python
dongdaemun_avg = aptRenamed.loc[aptRenamed["gu"] == "동대문구", "salesPrice"].mean()
seongbuk_avg = aptRenamed.loc[aptRenamed["gu"] == "성북구", "salesPrice"].mean()

aptRenamed.loc[aptRenamed["gu"] == "동대문구", "meanPrice"] = dongdaemun_avg
aptRenamed.loc[aptRenamed["gu"] == "성북구", "meanPrice"] = seongbuk_avg

print(aptRenamed.loc[aptRenamed["gu"].isin(["동대문구", "성북구"]), ["gu", "salesPrice", "meanPrice"]].head())
```

- 2-7. 시(si), 자치구(gu), 행정동(dong) 열을 결합하여 새로운 열에 저장하는 코드를 작성
하시오. 결합된 결과를 출력하면 아래와 같은 형태의 문자열 벡터가 생성됩니다.

```python
aptRenamed["fullAddress"] = aptRenamed["si"] + " " + aptRenamed["gu"] + " " + aptRenamed["dong"]
print(aptRenamed["fullAddress"])
```
