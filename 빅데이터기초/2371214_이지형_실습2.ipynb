- 위 지출 내역을 september_expenses라는 이름의 리스트 변수에 저장하고, 리스트

```python
#1-1

september_expenses = [ 300000,50000,70000,30000]
print(september_expenses)
```

- 10월부터 월세 400,000원이 추가되었고, 교통비가 60,000원으로 인상되었다. 월세
를 리스트의 마지막에 추가하고, 교통비를 수정한 후 변경된 리스트를 출력하는 코드를
작성하시오.

```python
#1-2

september_expenses.append(400000)
september_expenses[1] = 60000
print(september_expenses)
```

- 김빅데이터 학생은 불필요하다고 생각되는 도서비(30,000원) 항목을 삭제하려고
한다. 도서비 항목을 삭제한 후 리스트를 출력하는 코드를 작성하시오. 

```python
#1-3

idx = september_expenses.index(30000)
del september_expenses[idx]

print(september_expenses)
```

- september_expenses 리스트에서 앞에서 삭제한 도서비 항목이 제대로 삭제되었는
지 확인하고, 남아 있는 전체 지출 항목의 개수를 출력하는 코드를 작성하시오.


```python
#1-4

book_money = (30000 in september_expenses)
if(book_money == False):
    print("도서비 항목이 제대로 삭제되었습니다")
else:
    print("도서비 항목이 삭제되지 않았습니다")
    

print(september_expenses)
```

- september_expenses 리스트를 오름차순 정렬 후 출력하고, 이어서 내림차순 정렬
후 출력하는 코드를 작성하시오.

```python
#1-5

september_expenses.sort()
print(september_expenses)
september_expenses.sort(reverse=True)
print(september_expenses)

#주의:sort()메서드는 정렬결과를 반환하지 않는 메서드 이기 때문에 print로 바로 감싸서 출력하면 None이 찍힌다!
```

- 10월에 추가로 발생한 지출 내역(여행비 150,000원, 취미비 80,000원)을 += 연산

```python
#1-6

september_expenses+=[150000,80000]
print(september_expenses)
```

## 2번

- 위 준비물들을 각각 튜플로 만들고, 이를 묶어 items라는 이름의 리스트에 저장한
뒤 전체를 출력하는 코드를 작성하시오

```python
#2-1

items = [('음료',5000),('간식',3000),('장식품',30000)]
print(items)
```

- items 리스트에서 첫 번째 준비물(음료) 튜플을 가져와서 언패킹을 통해 name과
price 변수에 할당하고, ‘음료:가격’ 형식으로 출력하는 코드를 작성하시오.

```python
#2-2

name,price = items[0]
print(name+":"+str(price)+"(원)")
```

- 첫 번째 준비물(음료) 튜플의 가격을 6,000원으로 직접 수정하려고 시도하고, 실행
시 발생하는 오류 메시지를 확인하여 출력하는 코드를 작성하시오.

```python
#2-3

try:
    (items[0])[1] = 6000
except Exception as e:
    print("발생한 오류:",e)
```

- 가격을 올바르게 반영하기 위해서는 새 튜플을 만들어 리스트 원소를 수정해야 한
다. 음료 가격을 6,000원으로 수정하고, 변경된 리스트를 출력하는 코드를 작성하시오.

```python
#2-4

items[0] = ('음료',6000)
print(items)
```

- 새로운 준비물(‘현수막’, 20,000원)을 추가하고, 불필요한 ‘간식’ 품목은 삭제한 후
변경된 리스트를 출력하는 코드를 작성하시오. 

```python
#2-5

items.append(('현수막',20000))
del items[1]
print(items)
```

- items 리스트를 가격 기준 오름차순으로 정렬한 결과를 sorted_items 변수에 저장
하고, 정렬된 리스트를 출력하는 코드를 작성하시오.

```python
#2-6

sorted_items = sorted(items,key=lambda x:x[1])
print(sorted_items)
```
