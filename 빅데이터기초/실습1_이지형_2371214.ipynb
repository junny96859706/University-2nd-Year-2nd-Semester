## 2번

- 위에서 주어진 가사를 lyrics라는 변수에 문자열 형태로 저장하는 코드를 작성하고,
저장된 lyrics 변수의 값을 출력하여 확인하는 코드를 작성하시오.


```python
# 2-1

lyrics = "You, you are my universe and I just want to put you first"
print(lyrics)

```

-  lyrics 변수에 저장된 가사의 전체 길이(문자 수)와 단어 개수를 계산하여 출력하는
코드를 작성하시오.

```python
# 2-2

lyrics_len = len(lyrics)
word_len = len(lyrics.split())

print("전체길이:", lyrics_len)
print("단어길이:", word_len)
```

- lyrics 변수에서 "universe"라는 단어만 슬라이싱을 이용하여 추출하는 코드를 작성
하시오

```python
# 2-3

print(lyrics[16:25])
```

- lyrics 변수에 저장된 가사를 전부 소문자로 변환한 뒤, 'you'라는 단어가 총 몇 번
등장하는지 계산하여 출력하는 코드를 작성하시오.


```python
# 2-4

lower_lyrics = lyrics.lower()
print(lower_lyrics.count('you'))
```

-  lyrics 변수에 포함된 모든 'you'라는 단어를 'we'로 변경하여 출력하는 코드를 작성
하시오.

```python
# 2-5


new_lyrics = lower_lyrics.replace("you","we")
print(new_lyrics)
```

- lyrics 변수를 공백(' ') 기준으로 분할하여 새로운 리스트 변수에 저장한 뒤, 해당
리스트의 두 번째 단어와 마지막 단어를 각각 출력하는 코드를 작성하시오.


```python
# 2-6

list_lyrics = lyrics.split()
print("두번째 단어:" , list_lyrics[1])
print("마지막 단어:" , list_lyrics[-1])
```

- lyrics 변수를 공백(' ') 기준으로 분할하여 리스트를 만든 후, 해당 리스트에서 앞의
세 단어만 선택하여 하이픈(-)으로 연결된 하나의 문자열로 결합하여 출력하는 코드를
작성하시오.


```python
# 2-7

list_lyrics = lyrics.split()
first_to_three = list_lyrics[:3]

result = '-'.join(first_to_three)
print(result)
```
