## 1. Matplotlib 라이브러리

### 1-1. Matplotlib 개요

- Matplotlib은 파이썬에서 가장 널리 사용되는 2D 데이터 시각화 라이브러리임
    - John D. Hunter가 2003년에 개발함
    - 정적 이미지 형태의 다양한 그래프를 생성할 수 있으며 EDA(탐색적 데이터 분석), 보고서용 시각 자료 제작, 기계학습 결과 시각화 등에 폭넓게 활용됨
    - `matplotlib.pyplot` 모듈을 사용하여 손쉽게 다양한 시각화 구현 가능함

### 1-2. 주요 특징

| 특징 | 설명 |
|---|---|
| 다양한 형태의 시각화 제공 | 선 그래프(line), 막대 그래프(bar), 히스토그램(hist), 산점도(scatter), 파이 차트(pie), 박스플롯(boxplot) 등 다양한 형태의 시각화 제공 |
| 커스터마이징 가능 | 축(label, ticks), 범례(legend), 색상(color), 폰트(font), 스타일(style) 등 세부 조정 가능 |
| NumPy, Pandas와 연동 | NumPy 배열 또는 Pandas 시리즈/데이터프레임을 직접 입력 가능 |
| 다양한 출력 포맷 지원 | PNG, PDF, SVG, EPS 등 다양한 이미지 포맷으로 저장 가능 |
| 인터랙티브 모드 지원 | Jupyter Notebook 환경에서 `%matplotlib inline`, `%matplotlib notebook` 명령으로 그래프 출력 모드 제어 가능 |

### 1-3. 시각화 기본 절차

- `matplotlib.pyplot` 모듈은 MATLAB과 유사한 명령형 인터페이스를 제공함
  - `pyplot`은 자동으로 `Figure`(도화지)와 `Axes`(좌표축)를 생성하여 손쉽게 시각화 가능함
  - 단순 그래프에는 `pyplot`이 적합하지만 복잡한 다중 그래프(subplot 등)에는 객체지향(OOP) 방식 권장함
- 시각화 기본 절차
  - 데이터 준비
  - `pyplot` 모듈 임포트
  - 시각화 함수 호출(`plt.plot()`, `plt.bar()` 등)
  - 제목·라벨·범례 설정(`plt.title()`, `plt.xlabel()`, `plt.legend()` 등)
  - `plt.show()`로 그래프 출력

```python
# ! pip install matplotlib  # 필요시 주석 해제하여 설치

# 데이터 준비
x = [1, 2, 3, 4]
y = [10, 20, 25, 30]

# pyplot 모듈 임포트
import matplotlib.pyplot as plt

# 시각화 함수 호출
plt.plot(x, y)

# 제목/라벨/범례 설정
plt.title("Simple Line Plot")
plt.xlabel("X-axis")
plt.ylabel("Y-axis")

# plt.show() 호출
plt.show()
```

### 1-4. 자주 사용하는 그래프 종류

| 그래프 종류 | 함수 | 활용 목적 |
|---|---|---|
| 선 그래프(Line) | `plt.plot()` | 시계열 변화 추세 시각화 |
| 막대 그래프(Bar) | `plt.bar()`, `plt.barh()` | 범주형 데이터 비교 |
| 히스토그램(Histogram) | `plt.hist()` | 데이터 분포 확인 |
| 산점도(Scatter) | `plt.scatter()` | 변수 간 상관관계 시각화 |
| 파이 차트(Pie) | `plt.pie()` | 비율 및 구성비 표현 |
| 박스플롯(Boxplot) | `plt.boxplot()` | 이상치 탐색 및 분포 확인 |
| 히트맵(Heatmap) | `plt.imshow()` | 행렬형 데이터(예: 상관계수) 시각화 |

## 2. 히스토그램

### 2-1. 히스토그램 개요

- 히스토그램은 데이터의 분포 형태를 직관적으로 파악할 수 있는 대표적인 시각화 도구임
  - 데이터를 연속된 구간(급간, bin)으로 나누고, 각 구간에 속한 데이터의 빈도수(frequency)를 막대의 높이로 표현함
  - 히스토그램은 데이터의 중심 경향, 분산, 왜도(skewness), 이상치(outlier) 등을 파악하는 데 유용함
- 급간(bin) 선택의 중요성
  - 급간의 개수는 히스토그램의 해석 결과에 직접적인 영향을 미침
  - 급간이 너무 많으면 데이터가 과도하게 세분화되어 노이즈가 부각됨
  - 급간이 너무 적으면 데이터의 분포 특성이 왜곡됨
  - 일반적으로 데이터의 크기와 다양성에 따라 5개~15개 급간 사용을 권장함
  - 데이터 수가 많고 다양할수록 더 많은 급간(bin)을 설정하는 것이 유리함

### 2-2. 예제: 국가별 교통사고 사망률

- 아래의 데이터는 30개 국가의 1980년 중반 교통사고 사망률(인구 10만 명당)을 나타냄

```python
# Pandas 라이브러리 호출
import pandas as pd

# 예제 데이터 불러오기
traffic = pd.read_csv("deathrates.csv")

# 데이터프레임의 크기 출력
print("데이터프레임의 크기:", traffic.shape)

# 데이터프레임의 처음 5개 행 출력
traffic.head()
```

- 히스토그램은 `plt.hist()` 함수를 사용하여 빠르고 효율적으로 만들 수 있음

```python
# 히스토그램 생성
plt.hist(traffic["death"])

# 그래프 표시
plt.show()
```

- `plt.hist()` 함수는 구간 개수, 막대 색상 등 다양한 옵션을 분석 목적에 맞게 설정할 수 있음
  - `bins`: 급간 개수
  - `color`: 막대 색상
  - `alpha`: 투명도
  - `rwidth`: 막대 너비 비율
  - `edgecolor`: 테두리 색상
  - `linewidth`: 테두리 선 두께

```python
# 급간, 색상 등 옵션 지정
plt.hist(traffic["death"],
         bins=6,
         color="lightblue",
         alpha=0.7,
         rwidth=0.95,
         edgecolor="black",
         linewidth=0.5)

# 그래프 표시
plt.show()
```

### 2-3. 그래프 옵션 설정

- 모든 그래프는 색상이나 제목 등을 꾸미기 위한 다양한 인자를 사용할 수 있음
  - `plt.figure()`: 그래프 전체 틀 설정(크기, 해상도, 배경색 등)
  - `plt.title()`: 그래프 제목 설정(텍스트, 크기, 위치, 색상 등)
  - `plt.xlabel()` / `plt.ylabel()`: 축 이름 지정
  - `plt.ylim()` / `plt.xlim()`: 축 범위 설정

```python
# 그래프 크기 설정
plt.figure(figsize=(10, 5),
           dpi=100,
           facecolor='white',
           edgecolor='black')

# 급간, 색상 등 인자 지정
plt.hist(traffic["death"],
         bins=6,
         color="lightblue")

# 제목 설정
plt.title('Traffic Death Rates',
          fontsize=18,
          fontweight='bold', pad=20)

# x축, y축 레이블 설정
plt.xlabel('Death Rates',
           fontsize=14,
           fontweight='bold',
           labelpad=10,
           color='black')

plt.ylabel('Frequency',
           fontsize=14,
           fontweight='bold',
           labelpad=10,
           color='black')

# y축 범위 설정
plt.ylim(0, 20)

# 그래프 표시
plt.show()
```

- Matplotlib에서는 색상을 이름, 16진수 코드(#RRGGBB) 등 다양한 방식으로 지정 가능함
  - 색상은 선(line), 텍스트(text), 축(axis), 배경(background) 등 그래프의 모든 요소에 적용 가능함
  - `matplotlib.colors` 모듈의 `CSS4_COLORS` 사전에는 CSS4 표준 색상명과 해당 RGB 값이 정의되어 있음

```python
from matplotlib import colors

# 사용 가능한 모든 색상 이름 출력
for name, hex in colors.CSS4_COLORS.items():
    print(name, hex)
```

## 3. 막대그래프

### 3-1. 막대그래프 개요

- 막대그래프는 빈도 분포표를 시각적으로 표현하는 대표적인 그래프임
  - 이산형(discrete) 또는 범주형(categorical) 변수의 데이터 분포를 한눈에 비교할 때 유용함
  - 각 범주의 데이터 값(또는 빈도수)을 막대의 높이로 표현하여 범주 간 차이를 직관적으로 파악할 수 있음
  - 세로 방향(`plt.bar()`) 또는 가로 방향(`plt.barh()`) 그래프로 표현 가능함

### 3-2. 예제: 도시 인구 변화

- `popcities10.csv` 데이터 파일은 한국의 여러 도시들에 대한 2000년, 2005년, 2010년의 인구를 나타냄

```python
# 예제 데이터 불러오기
pop = pd.read_csv("popcities10.csv")

# 데이터프레임의 크기 출력
print(pop.shape)

# 데이터프레임의 처음 5개 행 출력
pop.head()
```

- 막대그래프는 `plt.bar()` 함수를 사용하여 쉽게 만들 수 있음
  - 예제에서는 인구 수를 기준으로 상위 10개 도시를 선택하고, 각 도시의 2010년 인구 데이터를 사용함

```python
# 그래프 크기 설정
plt.figure(figsize=(12, 8))

# 막대 그래프 생성
plt.bar(pop['municipality'][:10], pop['pop10'][:10])

# 그래프 표시
plt.show()
```

- y축의 값이 지나치게 클 경우에는 단위(예: 1,000명 단위로 환산)를 변경하여 가독성을 높이는 것이 효과적임

```python
# 그래프 크기 설정
plt.figure(figsize=(12, 8))

# y축 인구 수 단위를 변경
plt.bar(pop['municipality'][:10], pop['pop10'][:10] / 1000)

# 제목 및 축 라벨 추가
plt.title('Population Change in Selected Cities')
plt.xlabel('City')
plt.ylabel('Population (in thousands)')

# 그래프 표시
plt.show()
```

- 막대그래프는 세로 방향뿐 아니라 `plt.barh()` 함수를 통해 수평 막대그래프로도 표현 가능함

```python
# 그래프 크기 설정
plt.figure(figsize=(12, 8))

# 막대 그래프 생성
plt.barh(pop['municipality'][:10], pop['pop10'][:10] / 1000)

# 제목 및 축 라벨 추가
plt.title('Population Change in Selected Cities')
plt.xlabel('Population (in thousands)')
plt.ylabel('City')

# 그래프 표시
plt.show()
```

- 하나의 범주에 대해 여러 시점 데이터를 비교해야 할 경우에는 그룹 형태의 막대그래프로 시각화할 수 있음
  - 각 그룹마다 시점별로 막대를 나란히 배치하여 연도별 추세를 비교함

```python
import numpy as np

# 도시 선택
selected_cities = pop['municipality'][:10]
pop_2000 = pop['pop00'][:10] / 1000
pop_2005 = pop['pop05'][:10] / 1000
pop_2010 = pop['pop10'][:10] / 1000

# X축의 위치 설정
x = np.arange(len(selected_cities))
```

```python
# 그래프 크기 설정
plt.figure(figsize=(12, 8))

# 막대그래프 그리기
plt.bar(x, pop_2000, width=0.2, label='2000 years')
plt.bar(x + 0.2, pop_2005, width=0.2, label='2005 years')
plt.bar(x + 0.4, pop_2010, width=0.2, label='2010 years')

# X축에 도시 이름 표시
plt.xticks(x + 0.2, selected_cities)

# 범례 표시
plt.legend()

# 제목 및 축 라벨 추가
plt.title('Population Change in Selected Cities')
plt.xlabel('City')
plt.ylabel('Population (in thousands)')

# 그래프 표시
plt.show()
```

## 4. 선 그래프

### 4-1. 선 그래프 개요

- 선 그래프는 시간의 흐름에 따른 변화 추세나 경향을 시각화하는 데 사용되는 대표적인 그래프임
  - 일반적으로 연속형(continuous) 데이터, 특히 시계열(time series) 데이터의 변화를 표현하는 데 유용함
  - 연도별 변화, 누적 값 추세, 시간 흐름에 따른 데이터 패턴을 파악하는 데 효과적임
  - 데이터 포인트를 선(line)으로 연결하여 전반적인 흐름을 한눈에 파악할 수 있음

### 4-2. 예제: 도시 인구 변화

- 아래 예제 데이터는 한국의 여러 도시들에 대한 연도별 인구 변화를 나타내고 있음

```python
# 예제 데이터 불러오기
pop = pd.read_csv("popcities10.csv")

# 데이터프레임의 크기 출력
print(pop.shape)

# 데이터프레임의 처음 5개 행 출력
pop.head()
```

- `plt.plot()` 함수를 이용하여 선 그래프를 쉽게 생성할 수 있음

```python
# plt.plot() 함수
plt.plot([2000, 2005, 2010], pop.iloc[0, 3:6].sort_index() / 1000,
         marker='o',
         linestyle='dashed',
         color='r')

# 그래프 표시
plt.show()
```

### 4-3. 범례 설정

- 여러 도시의 선 그래프를 동시에 출력하면 도시간 인구 증가 추이를 직관적으로 비교할 수 있음
  - `label` 인자를 사용하여 도시명을 지정하고 `plt.legend()`를 활용하면 범례를 통해 그래프 해석을 쉽게 할 수 있음

```python
# 여러 선 그래프 생성
plt.plot([2000, 2005, 2010], pop.iloc[2, 3:6].sort_index() / 1000,
         marker='o',
         linestyle='dashed',
         color='b',
         label='Incheon')

plt.plot([2000, 2005, 2010], pop.iloc[3, 3:6].sort_index() / 1000,
         marker='o',
         linestyle='solid',
         color='b',
         label='Daegu')

plt.ylim(2000, 3200)

# 범례 추가
plt.legend(facecolor='white', edgecolor='black', loc='upper right')

# 그래프 표시
plt.show()
```

### 4-4. 마커 및 선 스타일 설정

- `plt.plot()` 함수를 사용할 때 `marker`와 `linestyle` 옵션을 활용하면 데이터 포인트의 표현 방식과 이들을 연결하는 선의 스타일을 다양하게 조정할 수 있음
  - `marker`: 데이터 포인트 각각에 적용될 마커(점 기호)의 유형을 지정
  - `linestyle`: 데이터 포인트를 연결하는 선의 스타일을 지정

```python
# 예시 데이터
x = [1, 2, 3, 4, 5]
y = [2, 3, 5, 7, 11]

# 산점도 및 선 그래프 생성
plt.plot(x, y, marker='o', linestyle='solid', label='Solid Line')
plt.plot(x, [i+1 for i in y], marker='s', linestyle='dotted', label='Dotted Line')
plt.plot(x, [i+2 for i in y], marker='^', linestyle='dashed', label='Dashed Line')
plt.plot(x, [i+3 for i in y], marker='*', linestyle='dashdot', label='Dash-dot Line')

# 범례 표시
plt.legend()

# 그래프 제목 및 축 라벨 설정
plt.title('Different Marker and Linestyle')
plt.xlabel('X axis')
plt.ylabel('Y axis')

# 그래프 표시
plt.show()
```

## 5. 산점도

### 5-1. 산점도 개요

- 산점도는 두 변수 간의 관계(상관관계)를 시각적으로 표현하는 데 적합한 그래프임
  - 각 데이터 포인트를 x축과 y축 값에 따라 점(point)으로 표시함
  - 변수 간의 방향성(양의/음의 관계), 선형성(linearity), 관계 강도(strength)를 직관적으로 파악 가능함
  - 데이터의 전반적 패턴에서 벗어난 이상치(outlier)도 쉽게 식별할 수 있음

### 5-2. 예제: 항공사 아웃소싱

- 아래 데이터는 각 항공사의 아웃소싱 비율과 그에 따른 운항 지연 비율을 나타냄

```python
# 예제 데이터 불러오기
flights = pd.read_csv("outsourcing.csv")

# 데이터프레임의 크기 출력
print(flights.shape)

# 데이터프레임의 처음 5개 행 출력
flights.head()
```

- `plt.scatter()` 함수를 사용하여 산점도를 생성할 수 있음
  - 마커 유형, 색상 및 크기 수정 등 다양한 옵션을 설정할 수 있음

```python
# scatter() 함수 사용
plt.scatter(flights["outsourcing"], flights["delays"],
            marker='o',
            s=20,
            c='r')

# 그래프 표시
plt.show()
```

### 5-3. 기준선 설정

- `plt.axhline()` 함수는 그래프에 수평선을 추가할 수 있는 도구임
  - 특정 기준선이나 임계값을 시각적으로 강조할 때 유용함
  - 선의 색상, 스타일, 두께 등 다양한 속성을 설정할 수 있음

```python
# 산점도 생성
plt.scatter(flights["outsourcing"], flights["delays"], marker='v')

# 제목 및 축 라벨 추가
plt.title('Outsourcing vs. Flight Delays')
plt.xlabel('Percentage of Outsourced Maintenance')
plt.ylabel('Percentage of Flight Delays')

# 수평선 추가
plt.axhline(y=20,
            color='b',
            linestyle='dashed',
            linewidth=1)

# 그래프 표시
plt.show()
```
