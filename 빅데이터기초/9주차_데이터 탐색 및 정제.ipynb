## 2. 시리즈 (Series)


### 2-1. 시리즈 개요

- 시리즈는 데이터가 순차적으로 나열된 1차원 배열 형태를 가짐
  - 인덱스(index)와 데이터 값(value)이 서로 매칭되는 구조임
  - 이런 관점에서 보면 키(key)-값(value) 형태로 대응되는 딕셔너리와 유사한 구조라고 볼 수 있음
  - 시리즈의 인덱스는 데이터 값의 위치를 나타내는 역할을 함


### 2-2. 시리즈 생성

- 딕셔너리와 시리즈의 구조가 비슷하기 때문에 딕셔너리를 시리즈로 변환하는 방식을 많이 사용함
- pandas의 내장 함수인 `Series()`에 딕셔너리를 인자로 전달하면 키(key)는 인덱스, 값(value)은 데이터 값으로 변환됨


```python
# 딕셔너리를 시리즈로 변환
sr_data = pd.Series({"a": 1, "b": 2, "c": 3})
print("Series 출력:")
print(sr_data)

# 자료형 확인
print("\n객체 타입:", type(sr_data))
```

### 2-3. 시리즈 인덱스

- pandas 시리즈에서 인덱스는 매칭되는 데이터 값의 순서와 주소를 저장하는 역할을 하며, 기본적으로 두 종류의 인덱스가 존재함
  - 정수형 위치 인덱스(integer position): `0`, `1`, `2` …와 같이 자동으로 부여되는 인덱스(기본값)
  - 사용자 정의 인덱스(index label): 사용자가 직접 지정하는 인덱스 이름


- 인덱스를 따로 지정하지 않으면 기본적으로 정수형 위치 인덱스가 자동으로 부여됨
  - `index` 속성: 인덱스 배열을 확인
  - `values` 속성: 데이터 값 배열만 확인


```python
# 리스트를 시리즈로 변환
sr_data = pd.Series(["2020-01-01", 3.14, "ABC", 100, True])

print("Series 출력:\n", sr_data)

# 인덱스 확인
print("\n인덱스:", sr_data.index)

# 값 확인
print("값:", sr_data.values)
```

### 2-4. 원소 선택

- 인덱스를 사용하면 시리즈의 원소를 선택할 수 있음
  - 하나의 원소만 선택하거나 여러 원소를 동시에 선택할 수 있음
  - 파이썬 리스트의 슬라이싱과 유사하게 인덱스 범위를 지정할 수도 있음
- 인덱스 사용 방법
  - 정수형 위치 인덱스: 대괄호 안에 숫자를 입력해 위치로 접근
  - 사용자 정의 인덱스: 대괄호 안에 인덱스 이름을 입력해 접근


- `Series()` 함수를 사용하면 튜플도 시리즈로 변환할 수 있음
  - 튜플은 리스트처럼 인덱스를 자동으로 가지지만, 시리즈로 변환할 때 사용자 지정 인덱스를 함께 전달하면 더 직관적인 데이터 구조를 만들 수 있음
- 앞서 설명한 대로 대괄호에 인덱스를 입력하여 원소에 접근 가능함


```python
# 튜플 데이터를 시리즈로 변환
tup_data = ("Alice", 24, "Seoul", True)
sr_data = pd.Series(tup_data, index=["name", "age", "city", "student"])

print("Series 출력:\n", sr_data)

# 인덱스를 사용한 원소 선택
# print("\n정수형 인덱스로 선택:", sr_data[0])  # KeyError 발생
print("인덱스 이름으로 선택:", sr_data["name"])
```

- 여러 개의 인덱스를 리스트 형태로 대괄호에 입력하면 해당 원소들을 한 번에 선택할 수 있음
  - 파이썬과 유사하게 슬라이싱 문법을 사용하여 인덱스 범위를 지정해 선택할 수도 있음


```python
# 여러 개 선택 (정수형 인덱스)
# print("정수형 인덱스 여러 개 선택:\n", sr_data[[1, 2]])  # KeyError 발생

# 슬라이싱 (정수형 위치 기반)
print("\n슬라이싱 - 정수형 인덱스:\n", sr_data[1:3])

# 여러 개 선택 (인덱스 이름)
print("\n인덱스 이름 여러 개 선택:\n", sr_data[["name", "age"]])
```

## 3. 데이터프레임 (DataFrame)


### 3-1. 데이터프레임 개요

- 데이터프레임은 행과 열로 구성된 2차원 배열 구조를 나타냄
- 데이터프레임은 여러 개의 시리즈가 동일한 행 인덱스를 기준으로 결합된 형태라고 볼 수 있으며, 2차원 벡터 또는 행렬(matrix) 구조로 이해할 수 있음


### 3-2. 데이터프레임 생성

- 데이터프레임을 만들기 위해서는 같은 길이(원소 개수가 동일한)의 1차원 배열 여러 개가 필요함
  - 데이터프레임은 여러 시리즈를 모아 놓은 집합이며, 딕셔너리의 값에 해당하는 각 리스트가 시리즈 배열로 변환되어 열을 구성함
  - 딕셔너리의 키는 각 시리즈의 이름으로 변환되어 데이터프레임의 열 이름으로 지정됨


```python
# 딕셔너리를 사용한 데이터프레임 생성
data = {
    "col1": [1, 2, 3],
    "col2": [4, 5, 6],
}
df = pd.DataFrame(data)

print("DataFrame 출력:\n", df)

# 데이터프레임 타입 확인
print("\n객체 타입:", type(df))
```

- 데이터프레임은 2차원 배열을 쉽게 변환할 수 있는 구조적 특성을 가지고 있음
- 행 인덱스(`index` 속성)와 열 이름(`columns` 속성)을 직접 지정할 수 있음


```python
# 2차원 리스트를 데이터프레임으로 변환
data = [
    [101, "Alice", "Seoul"],
    [102, "Bob", "Busan"],
]

df = pd.DataFrame(
    data,
    index=["s1", "s2"],             # 행 인덱스 지정
    columns=["ID", "Name", "City"]  # 열 이름 지정
)

print("DataFrame 출력:\n", df)

# 행 인덱스와 열 이름 확인
print("\n행 인덱스:", df.index)
print("열 이름:", df.columns)
```

## 4. 데이터 탐색


### 4-1. 파일 읽어오기

- pandas는 다양한 형태의 외부 파일을 읽어와 데이터프레임으로 변환하는 함수를 제공함
  - 어떤 파일이든 데이터프레임으로 변환되면 pandas의 모든 기능을 동일하게 활용할 수 있음
  - 반대로 데이터프레임을 다양한 파일 형식으로 저장할 수도 있음

| File format | Reader 함수 | Writer 함수 |
|---|---|---|
| CSV | `read_csv` | `to_csv` |
| JSON | `read_json` | `to_json` |
| HTML | `read_html` | `to_html` |
| Clipboard | `read_clipboard` | `to_clipboard` |
| Excel | `read_excel` | `to_excel` |
| HDF5 | `read_hdf` | `to_hdf` |
| SQL | `read_sql` | `to_sql` |


### 4-2. CSV 파일

- `pd.read_csv()` 함수에 파일 경로(파일명)를 입력하면 CSV 파일을 읽어와 데이터프레임으로 변환함
  - `header`: 데이터프레임의 열 이름으로 사용할 행 지정
  - `index_col`: 데이터프레임의 행 인덱스가 될 열 지정


```python
# CSV 파일 읽기 (특정 열을 인덱스로 지정)
exam_index = pd.read_csv("csv_exam.csv", index_col="id")
print(exam_index.head(3))  # 상위 3행 출력
```

```python
# CSV 파일 읽기 (기본 사용)
exam = pd.read_csv("csv_exam.csv", header=0)  # header=0(기본값)은 첫 행을 열 이름으로 사용
print(exam.head(3))  # 상위 3행 출력
```

- CSV 파일은 기본적으로 쉼표(`,`)로 구분되지만, 경우에 따라 탭(`\t`)이나 공백(`" "`)으로 구분되기도 함
  - 이 경우 `sep`(또는 `delimiter`) 옵션을 통해 구분 문자를 지정해야 함

| 옵션 | 설명 |
|---|---|
| `path` | 파일의 경로 |
| `sep` 또는 `delimiter` | 텍스트 데이터를 필드별로 구분하는 문자(예: `","`, `"\t"`) |
| `header` | 열 이름으로 사용할 행 번호 |
| `index_col` | 행 인덱스로 사용할 열 번호 또는 열 이름 |
| `names` | 열 이름으로 사용할 문자열 리스트 |
| `skiprows` | 처음 몇 줄을 건너뛸지 설정 |
| `parse_dates` | 날짜 텍스트를 `datetime64`로 변환할지 설정 |
| `skipfooter` | 마지막 몇 줄을 건너뛸지 설정 |
| `encoding` | 텍스트 인코딩 방식 지정(예: `"utf-8"`, `"cp949"`) |


### 4-3. 데이터 확인

- `head()` 함수는 데이터프레임의 앞부분 일부 행을 출력하며, `tail()` 함수는 마지막 부분 일부 행을 출력함
  - 인자로 정수 `n`을 넣으면 앞/뒤에서 `n`개의 행을 확인할 수 있음(기본값: `n = 5`)


```python
# 앞부분 확인 (상위 4행)
print("앞부분 데이터:")
print(exam.head(4))

# 뒷부분 확인 (하위 2행)
print("\n뒷부분 데이터:")
print(exam.tail(2))
```

- `shape` 속성: 데이터프레임의 행(row) 개수와 열(column) 개수를 튜플 형태로 반환
- `info()` 메서드: 데이터프레임의 전체 구조 요약 정보를 출력


```python
# 데이터프레임의 크기 확인
print("데이터프레임 shape:", exam.shape)

# 데이터프레임 구조 요약 정보
print("\n데이터프레임 info:")
print(exam.info())
```

- `describe()` 함수: 데이터프레임에서 숫자형 데이터가 포함된 열의 주요 기초 통계 정보를 요약하여 출력
  - 기본적으로 숫자형 데이터만 요약하지만, 문자열 데이터 등 모든 열을 포함하려면 `include="all"` 옵션을 추가할 수 있음


```python
# 숫자형 데이터 통계 요약
print("숫자형 데이터 describe:")
print(exam.describe())
```

- `count()` 함수: 데이터프레임 각 열의 결측치가 아닌 값의 개수를 반환
- `value_counts()` 함수: 시리즈 객체에서 고유값의 빈도를 반환
  - 기본적으로 숫자형 데이터만 요약하지만, 문자열 데이터 등 모든 열을 포함하려면 `include="all"` 옵션을 추가할 수 있음


```python
# 열별 데이터 개수 확인
print("열별 데이터 개수:")
print(exam.count())

# 특정 열(class)의 고유값 개수 확인
print("\n'class' 열의 고유값 분포:")
print(exam["class"].value_counts())
```

## 5. 데이터 정제


### 5-1. 행 인덱스/열 이름 수정

- 데이터프레임의 행 인덱스(`index`)와 열 이름(`columns`)은 새롭게 지정하여 변경 가능
  - `rename()` 함수로 일부 인덱스나 열 이름을 선택적으로 수정할 수 있음
  - `inplace=True` 옵션을 주면 새로운 데이터프레임을 만들지 않고 기존 객체를 직접 수정함


```python
# 열 이름 수정: 'math' → 'Math'
df_renamed = exam.rename(columns={"math": "Math"})
print("열 이름 수정:")
print(df_renamed.head())

# 행 인덱스 수정: id 1 → 'S1', id 2 → 'S2'
df_index_renamed = exam.rename(index={1: "S1", 2: "S2"})
print("\n행 인덱스 수정:")
print(df_index_renamed.head())
```

### 5-2. 행/열 삭제

- `drop()` 함수는 데이터프레임에서 특정 행 또는 열을 제거할 때 사용함
  - 기본적으로 `index` 속성은 행, `columns` 속성은 열을 삭제함
  - 여러 개의 행이나 열을 동시에 삭제하려면 리스트 형태로 지정함


```python
# 행 삭제: 인덱스가 1인 행 제거
print("행 삭제 예시:")
print(exam.drop(index=1))

# 열 삭제: 'class' 열 제거
print("\n열 삭제 예시:")
print(exam.drop(columns=["class"]))
```

### 5-3. 행 선택

- 전체 데이터를 분석할 수도 있지만, 특정 행만 추출하여 분석하는 경우가 많으며, pandas는 두 가지 방식으로 행을 선택할 수 있음
  - `loc`: 인덱스 이름(index label) 기반 선택
  - `iloc`: 정수형 위치(integer position) 기반 선택


- `loc` 함수는 인덱스 이름으로 행을 선택하며, `iloc` 함수는 정수 위치(`0`부터 시작)로 행을 선택함
- 예: `exam` 데이터에서 `id`가 `5`인 행을 선택하고 싶을 때
  - `loc[5]`: 인덱스 값이 `5`인 행 반환
  - `iloc[4]`: 5번째 행(`0`부터 세었을 때 `index=4`) 반환


```python
# loc: 인덱스 이름(=5)으로 선택
print("loc 예시:")
print(exam_index.loc[5])

# iloc: 다섯 번째 행(=index 위치 4) 선택
print("\niloc 예시:")
print(exam_index.iloc[4])
```

- `loc`과 `iloc` 모두 리스트 형태로 여러 개의 행을 한 번에 선택할 수 있음


```python
# loc: 인덱스 이름 [1, 4, 5] 선택
print("loc 예시:")
print(exam_index.loc[[1, 4, 5]])

# iloc: 정수 위치 [0, 3, 4] 선택
print("\niloc 예시:")
print(exam_index.iloc[[0, 3, 4]])
```

- 슬라이싱을 이용하면 연속된 범위의 행을 한 번에 선택할 수 있음
  - `loc`: 인덱스 이름 범위 선택(끝 값 포함)
  - `iloc`: 정수 위치 범위 선택(끝 값 미포함)


```python
# loc: 인덱스 이름으로 1~3번까지 선택 (끝 값 포함)
print("loc[1:3] 결과:")
print(exam_index.loc[1:3])

# iloc: 정수 위치 0~2까지 선택 (끝 값 미포함)
print("\niloc[0:3] 결과:")
print(exam_index.iloc[0:3])
```

- 특정 열에서 주어진 조건을 만족하는 행만 선택할 수 있음
  - 여러 조건을 동시에 사용할 때는 `&`(and), `|`(or)를 활용하고 반드시 괄호로 묶어야 함


```python
# class가 1인 행 선택
print("class == 1")
print(exam_index[exam_index["class"] == 1])

# class가 1이고, math 점수가 50 이상인 행 선택 (and 조건)
print("\n(class == 1) & (math >= 50)")
print(exam_index[(exam_index["class"] == 1) & (exam_index["math"] >= 50)])
```

- `isin()` 함수는 특정 열에서 여러 개의 값 중 하나라도 일치하는 경우를 찾아줌
  - SQL의 `IN` 조건과 유사하며, 리스트, 튜플, 시리즈를 인자로 받을 수 있음


```python
# 수학 점수가 80 또는 90인 행만 선택
print("수학 점수가 80 또는 90인 행만 선택")
print(exam_index[exam_index["math"].isin([80, 90])])

# 영어 점수가 78, 89인 행만 선택
print("\n영어 점수가 78, 89인 행만 선택")
print(exam_index[exam_index["english"].isin([78, 89])])
```

### 5-4. 열 선택

- 데이터프레임에서 특정 열만 추출할 때는 대괄호(`[]`) 또는 `loc`를 사용함
  - 여러 개의 열을 선택하려면 리스트(`[]`)로 묶어서 지정함


- 대괄호 안에 열 이름을 넣어 데이터프레임에서 일부 열만 선택 가능함
  - 단일 열 선택: 시리즈 반환
  - 다중 열 선택: 데이터프레임 반환


```python
# 1개 열 선택 (Series 반환)
print("1개 열 선택\n", exam["math"].head(3))
print("\n1개 열 선택 → Series 반환:", type(exam["math"]))

# 2개 이상 열 선택 (DataFrame 반환)
print("\n2개 이상 열 선택\n", exam.loc[:, ["math", "english"]].head(3))
print("\n2개 이상 열 선택 → DataFrame 반환:", type(exam.loc[:, ["math", "english"]]))
```

### 5-5. 원소 선택

- 행 인덱스와 열 이름을 함께 지정하여 원하는 원소(셀 데이터)에 접근할 수 있음
- 기본 문법: `df.loc[행, 열]` 또는 `df.iloc[행, 열]`
  - 1개 행 + 1개 열 → 스칼라 값
  - 1개 행 + 여러 열, 여러 행 + 1개 열 → 시리즈
  - 여러 행 + 여러 열 → 데이터프레임


- `loc[행인덱스, 열이름]`: 인덱스 이름 기반 접근
- `iloc[행위치, 열위치]`: 정수 위치 기반 접근
  - 두 방법 모두 원소 1개를 선택하면 스칼라 값이 반환됨


```python
# loc 예제: 행 인덱스=1, 열 이름='math'
print("행 인덱스=1, 열 이름='math'")
print(exam.loc[1, "math"])

# iloc 예제: 첫 번째 행(0), 두 번째 열(1)
print("\n첫 번째 행(0), 두 번째 열(1)")
print(exam.iloc[0, 1])
```

- 리스트 형태로 열/행을 지정하면 원하는 여러 개 원소를 동시에 선택할 수 있음
  - 슬라이싱(`:`)을 사용하면 연속된 범위의 원소를 쉽게 선택할 수 있으며, 반환되는 객체는 시리즈임


```python
# loc: 행 인덱스=1, 여러 열 선택
print("행 인덱스=1, 여러 열 선택")
print(exam.loc[1, ["math", "english", "science"]])

# iloc: 0번째 행, 열 위치 범위(1~3)
print("\n0번째 행, 열 위치 범위(1~3)")
print(exam.iloc[0, 1:4])
```

- 행 인덱스와 열 이름(또는 위치)을 동시에 지정해 데이터프레임 일부를 선택 가능함
  - 행과 열 모두 2개 이상 선택하면 데이터프레임을 반환함
  - 행과 열 중 하나만 여러 개 선택하면 시리즈를 반환함


```python
# loc: 행 인덱스 1~4, 열 이름 3개 선택
print("행 인덱스 1~4, 열 이름 3개 선택")
print(exam.loc[1:4, ["math", "english", "science"]])

# iloc: 행 위치 0~3, 열 위치 1~3
print("\n행 위치 0~3, 열 위치 1~3")
print(exam.iloc[0:4, 1:4])
```

### 5-6. 파생변수

- 기존 데이터프레임에 새로운 열을 만들어 추가 가능함
  - 파생 변수는 기존 변수들의 연산 결과를 활용하여 생성함


- 데이터프레임에서 새로운 열을 추가하여 파생변수 생성 가능
  - 하나의 값(스칼라)을 넣으면 모든 행에 동일하게 채워짐
  - 리스트나 시리즈를 넣으면 행 단위로 각각 값이 들어감


```python
# 모든 행에 동일한 값 추가
exam["music"] = 90

# 기존 열을 활용한 파생변수 추가 (총점)
exam["total"] = exam["math"] + exam["english"] + exam["science"]

print(exam.head())
```

- 데이터프레임에서는 특정 열을 선택하여 새로운 파생변수를 만들 수 있음
  - `sum()`, `mean()`과 같은 집계 함수를 활용해 행 단위로 합계 혹은 평균을 계산 가능함


```python
# 평균 파생변수 생성
exam["mean"] = exam[["math", "english", "science", "music"]].mean(axis=1)

print(exam.head())
```

- `apply()`와 `lambda`를 이용해 조건에 따라 새로운 값을 넣을 수 있음
  - `np.where()`를 이용하면 더 직관적으로 조건 분류 가능함


```python
# lambda + apply
exam["lambda"] = exam["total"].apply(lambda x: "Pass" if x > 60 else "Fail")

# np.where
import numpy as np
exam["where"] = np.where(exam["mean"] > 60, "Pass", "Fail")

print(exam.head())
```

### 5-7. 데이터 정렬

- `sort_values()` 함수를 사용하면 특정 열을 기준으로 정렬 가능함


- `ascending=True`(기본값): 오름차순 정렬
- `ascending=False`: 내림차순 정렬


```python
# 오름차순 정렬
print("평균(mean) 기준 오름차순 정렬 상위 3개 출력")
print(exam.sort_values("mean").head(3))

# 내림차순 정렬
print("\n평균(mean) 기준 내림차순 정렬 상위 3개 출력")
print(exam.sort_values("mean", ascending=False).head(3))
```
