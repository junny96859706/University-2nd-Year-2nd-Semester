## 1번

- 앞서 정의한 fahr2celsius( ) 함수를 활용하여 주어진 화씨온도를 절대온도(K)로 변
환하여 반환하는 함수 fahr2kelvin( )을 정의하는 코드를 작성하시오.


```python
#1-1
#화씨단위를 섭씨 단위로 변환하는 함수
def fahr2celsius(temp):
    celsius = (temp - 32) * (5 / 9)
    return celsius

#화씨온도를 절대온도(k)로 변환하여 반환하는 함수
def fahr2kelvin(temp):
    celsius = fahr2celsius(temp)
    return (celsius + 273.15)
```

- 주어진 섭씨 단위의 온도를 화씨 단위로 변환하는 함수 celsius2fahr( )를 정의하는
코드를 작성하시오.

```python
#1-2
#섭씨단위를 화씨단위로 변환하는 함수
def celsius2fahr(temp):
    fahr = (temp * 1.8) + 32
    return fahr
```

- 주어진 섭씨온도를 화씨온도와 절대온도로 각각 변환하여 하나의 딕셔너리로 반환
하는 함수 cel2FK( )를 정의하는 코드를 작성하시오. 반환되는 딕셔너리에는 화씨온도와
절대온도가 각각 "fahrenheit"와 "kelvin"이라는 두 개의 키로 설정하고, 각 키에 대응하
는 값으로는 변환된 화씨 온도와 절대온도가 저장되도록 설정하시오.

```python
#1-3
def cel2FK(temp):
    fahrenheit = celsius2fahr(temp)
    kelvin = temp + 273.15
    return {"fahrenheit":fahrenheit, "kelvin":kelvin}
```

## 2번

- 주어진 정보를 참고하여 무작위로 n명의 사람을 선택했을 때 최소 두 사람의 생일이 같을
확률을 계산하는 함수 bday( )를 정의하는 코드를 작성하시오. 

```python
import math

def bday(n):
    list_prod = []
    for i in range(n):
        list_prod.append((365-i)/365)
    return 1 - math.prod(list_prod)

print(f"{bday(2):.9f}")
```
