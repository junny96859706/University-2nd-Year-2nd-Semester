- 1-1. libraries.csv 파일을 읽어오고, 이를 libs라는 이름의 데이터프레임 객체로 저장하는
코드를 작성하시오.

```python
%pip install pandas
import pandas as pd

libs = pd.read_csv("libraries.csv", header=0)
```

- 1-2. libs 객체 내의 모든 열 이름을 출력하는 코드를 작성하시오

```python
print("libs 객체 내의 모든 열 이름:")
print(libs.columns)
```

- 1-3. libs 객체에 포함된 모든 도서관의 열람좌석수를 합산하는 코드를 작성하시오.

```python
print("열람좌석수:", libs["열람좌석수"].sum())
```

- 1-4. libs 객체에서 열람좌석수가 30석 이상인 도서관의 이름을 출력하는 코드를 작성하
시오

```python
print("열람좌석수가 30석 이상인 도서관의 이름:")
print(libs.loc[libs["열람좌석수"] >= 30, "도서관명"])
```

- 1-5. libs 객체에서 자료(도서) 수가 5,000권 이상이거나 건물면적이 100㎡ 이상인 도서
관의 모든 정보를 출력하는 코드를 작성하시오.

```python
print(libs[(libs["자료수(도서)"] >= 5000) | (libs["건물면적"] >= 100)])
```

- 1-6.  건물면적 1㎡당 열람좌석수가 0.3석 이상인 도서관의 도로명 주소를 출력하는 코드를 작성하시오

```python
print(libs.loc[(libs["열람좌석수"] / libs["건물면적"] >= 0.3), "소재지도로명주소"])
```

# 2번

- 2-1.  popcities15.csv 파일을 읽어오고, pop이라는 이름의 데이터프레임 객체로 저장하는
코드를 작성하시오

```python
pop = pd.read_csv("popcities15.csv", header = 0)
```

- 2-2. pop 객체 내 각 열의 데이터 타입을 확인하는 코드를 작성하고, 주석을 사용하여
각 열의 데이터 타입을 작성하시오.

```python
print("pop 객체 내 각 열의 데이터 타입:")
print(pop.dtypes)

# rank: int64
# provinces: str
# municipality: str
# pop15: int64
# pop10: int64
# pop05: int64
# pop00: int64
```

- 2-3. pop 객체의 홀수 번째 행을 모두 출력하는 코드를 작성하시오.

```python
print(pop.iloc[::2])
```

- 2-4. 2015년 인구(pop15)가 100만 명 이상인 도시명을 모두 선택하여 출력하는 코드를
작성하시오.

```python
print(pop.loc[pop["pop15"] >= 1000000, "municipality"])
```

- 2-5. 2000년 인구가 30만 이하이고, 2015년 인구가 40만 이상으로 증가한 도시명을 선
택하여 출력하는 코드를 작성하시오.

```python
print(pop.loc[(pop["pop00"] <= 300000) & (pop["pop15"] >= 400000), "municipality"])
```

- 2-6. 경기도에서 2010년 기준 인구가 70만 이상인 도시명을 선택하여 출력하는 코드를
작성하시오.

```python
print(pop.loc[(pop["provinces"] == "Gyeonggi") & (pop["pop10"] >= 700000), "municipality"])
```

- 2-7. 2005년에서 2010년 사이 인구 증가폭이 5만 명 이상인 도시명을 선택하여 출력하
는 코드를 작성하시오.


```python
print(pop.loc[((pop["pop10"] - pop["pop05"]) >= 50000), "municipality"])
```

- 2-8. 2000년에 비해 2015년 인구가 50% 이상 증가한 도시의 수를 출력하는 코드를 작
성하시오.


```python
print("50%이상 증가한 도시의 수:")
print(len(pop.loc[(pop["pop15"] >= pop["pop00"] * 1.5), "municipality"]))
```
